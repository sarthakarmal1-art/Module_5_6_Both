## 1. What is a CSV file? (8 min)

CSV means **comma-separated values**. Each line is a row and commas separate fields. A CSV stores values and headers, but not formulas, colours, or multiple sheets. Its simplicity makes it useful for exchange between software.

A `.csv` file can be opened in Excel, but treat it as data—not as a formatted report.

## 4. Common import problems (12 min)

- Wrong folder or filename → check the Explorer panel and spelling.
- Separator is `;` rather than comma → try `pd.read_csv('file.csv', sep=';')`.
- Decimal commas (`12,5`) → agree on a format or use suitable import options.
- Extra title rows → use `skiprows=` only after checking the source.
- Text where numbers should be → inspect and clean rather than forcing a calculation.

The most important habit: do not silently assume an import is correct.
